# DMV Practical 5 – Air Quality Data Analysis

**Aim:** Explore Delhi air-quality data, inspect AQI and pollutant trends, create plots and distributions, and study correlations between pollutants and AQI.

Uses the Delhi AQI dataset URL referenced in the PDF. A CSV upload fallback is included.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
url = "https://raw.githubusercontent.com/cp99/India-Air-Quality-Dataset/main/Delhi_AQI_Dataset.csv"
try:
    df = pd.read_csv(url)
except Exception:
    print("Upload Delhi_AQI_Dataset.csv if the URL is unavailable.")
    from google.colab import files
    uploaded = files.upload()
    df = pd.read_csv(next(iter(uploaded)))

print("Dataset loaded:", df.shape)
display(df.head()); display(df.tail())
print(df.dtypes)
df.info()
display(df.describe())
```

```python
# Clean and prepare columns
df.columns = df.columns.str.strip()
df = df.loc[:, ~df.columns.str.startswith("Unnamed")]
df["AQI"] = pd.to_numeric(df["AQI"], errors="coerce")
date_col = "Date" if "Date" in df.columns else "date"
if date_col in df.columns:
    df[date_col] = pd.to_datetime(df[date_col], dayfirst=True, errors="coerce")

for col in df.select_dtypes(include=np.number).columns:
    df[col] = df[col].fillna(df[col].median())
df = df.dropna(subset=["AQI"])
print("Missing values:\n", df.isnull().sum())
display(df.head())
```

```python
# AQI and pollutant trends
plt.figure(figsize=(8,3))
plt.plot(df["AQI"], linewidth=1)
plt.title("AQI Trend Over Time"); plt.xlabel("Time"); plt.ylabel("AQI")
plt.grid(alpha=0.3); plt.show()

pollutants = [c for c in ["PM2.5","PM10","NO2","SO2","CO","O3"] if c in df.columns]
for col in pollutants:
    plt.figure(figsize=(7,3))
    plt.plot(df[col], label=col)
    plt.title(col + " Trend"); plt.xlabel("Time"); plt.ylabel(col)
    plt.legend(); plt.grid(alpha=0.3); plt.show()

if pollutants:
    df[pollutants].plot(figsize=(9,4), title="Pollutant Trends Over Time")
    plt.xlabel("Time"); plt.ylabel("Pollutant Level"); plt.tight_layout(); plt.show()
```

```python
# AQI distributions and pollutant comparisons
plt.figure(figsize=(6,3))
plt.hist(df["AQI"], bins=30)
plt.title("AQI Distribution"); plt.xlabel("AQI"); plt.ylabel("Frequency")
plt.show()

if pollutants:
    df[pollutants].plot(kind="box", figsize=(7,4), title="Pollutant Distribution")
    plt.ylabel("Concentration"); plt.show()

for col in pollutants:
    plt.figure(figsize=(5,3))
    plt.scatter(df[col], df["AQI"], alpha=0.5)
    plt.xlabel(col); plt.ylabel("AQI"); plt.title("AQI vs " + col)
    plt.grid(alpha=0.3); plt.show()
```

```python
# Date-wise summary and correlation
if date_col in df.columns:
    daily_aqi = df.groupby(df[date_col].dt.date)["AQI"].mean()
    daily_aqi.plot(figsize=(9,3), title="Average AQI by Date")
    plt.xlabel("Date"); plt.ylabel("Average AQI"); plt.tight_layout(); plt.show()

numeric_cols = df.select_dtypes(include=np.number)
correlation = numeric_cols.corr()
display(correlation)
plt.figure(figsize=(8,6))
sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap"); plt.tight_layout(); plt.show()
```
