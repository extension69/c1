# DMV Practical 1 – Data Wrangling and Sales Analysis

**Aim:** Create sales data, export it to CSV/Excel/JSON, load the files, clean and combine the data, and perform sales analysis with visualizations.

The practical uses a small sales dataset defined in the notebook, so no external CSV is required.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import json
from IPython.display import display

sales_data = {
    "Order_ID": [1001,1002,1003,1004,1005,1006,1007,1008,1009,1010],
    "Date": ["2026-01-05","2026-01-10","2026-01-15","2026-02-02","2026-02-12",
             "2026-02-20","2026-03-03","2026-03-15","2026-03-20","2026-03-25"],
    "Product": ["Laptop","Mobile","Tablet","Laptop","Mobile","Headphones","Tablet","Laptop","Mobile","Headphones"],
    "Category": ["Electronics","Electronics","Electronics","Electronics","Electronics",
                 "Accessories","Electronics","Electronics","Electronics","Accessories"],
    "Region": ["North","South","East","West","North","South","East","West","North","South"],
    "Quantity": [2,5,3,1,4,6,2,3,5,4],
    "Unit_Price": [60000,25000,30000,60000,25000,2000,30000,60000,25000,2000]
}
df = pd.DataFrame(sales_data)
df["Sales"] = df["Quantity"] * df["Unit_Price"]
display(df)
```

```python
# Export to CSV, Excel and JSON
df.to_csv("sales_data.csv", index=False)
df.to_excel("sales_data.xlsx", index=False)
df.to_json("sales_data.json", orient="records", indent=4)
print("CSV, Excel and JSON files created.")
```

```python
# Load the saved files
csv_data = pd.read_csv("sales_data.csv")
excel_data = pd.read_excel("sales_data.xlsx")
json_data = pd.read_json("sales_data.json")

print("CSV Data"); display(csv_data.head())
print("Excel Data"); display(excel_data.head())
print("JSON Data"); display(json_data.head())
```

```python
# Inspect and clean data
print(csv_data.info())
print("\nMissing values:\n", csv_data.isnull().sum())
print("\nDuplicate rows:", csv_data.duplicated().sum())

cleaned_data = csv_data.drop_duplicates().copy()
cleaned_data["Date"] = pd.to_datetime(cleaned_data["Date"])
for col in ["Product", "Category", "Region"]:
    cleaned_data[col] = cleaned_data[col].str.strip().str.title()
cleaned_data["Quantity"] = cleaned_data["Quantity"].fillna(cleaned_data["Quantity"].median())
cleaned_data["Unit_Price"] = cleaned_data["Unit_Price"].fillna(cleaned_data["Unit_Price"].median())
cleaned_data["Sales"] = cleaned_data["Quantity"] * cleaned_data["Unit_Price"]
display(cleaned_data)
```

```python
# Combine datasets and create date features
unified_data = pd.concat([csv_data, excel_data, json_data], ignore_index=True).drop_duplicates()
unified_data["Date"] = pd.to_datetime(unified_data["Date"])
unified_data["Month"] = unified_data["Date"].dt.month
unified_data["Month_Name"] = unified_data["Date"].dt.month_name()
unified_data["Year"] = unified_data["Date"].dt.year
unified_data["Sales"] = unified_data["Quantity"] * unified_data["Unit_Price"]
display(unified_data)
print("Unified data shape:", unified_data.shape)
```

```python
# Summary statistics and sales analysis
print("Total Sales:", unified_data["Sales"].sum())
print("Average Order Value:", round(unified_data["Sales"].mean(), 2))
print("Total Quantity Sold:", unified_data["Quantity"].sum())

product_sales = unified_data.groupby("Product")["Sales"].sum().sort_values(ascending=False)
region_sales = unified_data.groupby("Region")["Sales"].sum().sort_values(ascending=False)
category_sales = unified_data.groupby("Category")["Sales"].sum().sort_values(ascending=False)
monthly_sales = unified_data.groupby("Month_Name")["Sales"].sum()
print("\nSales by Product:"); display(product_sales)
print("\nSales by Region:"); display(region_sales)
print("\nSales by Category:"); display(category_sales)
print("\nSales by Month:"); display(monthly_sales)
```

```python
# Visualizations
product_sales.plot(kind="bar", figsize=(6,3), title="Total Sales by Product", ylabel="Sales")
plt.tight_layout(); plt.show()

region_sales.plot(kind="bar", figsize=(6,3), title="Total Sales by Region", ylabel="Sales")
plt.tight_layout(); plt.show()

category_sales.plot(kind="bar", figsize=(6,3), title="Total Sales by Category", ylabel="Sales")
plt.tight_layout(); plt.show()
```
